```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")
assert api_key, "❌ OPENROUTER_API_KEY not found in .env!"
print(f"✅ API key loaded (starts with: {api_key[:12]}...)")

client = OpenAI(api_key=api_key, base_url="https://openrouter.ai/api/v1")
```

```python
models = {
"GPT-4o-mini":   "openai/gpt-4o-mini",
"Claude Sonnet": "anthropic/claude-sonnet-4.5",
"Gemini Flash":  "google/gemini-2.0-flash-001",
"DeepSeek V3":   "deepseek/deepseek-chat",
"Llama 3.3":     "meta-llama/llama-3.3-70b-instruct",
"Qwen 2.5":      "qwen/qwen-2.5-72b-instruct",
}

prompt = "Reply with exactly this and nothing else: API works!"

results = {}
for name, model_id in models.items():
    try:
        r = client.chat.completions.create(
            model=model_id,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=20,
        )
        reply = r.choices[0].message.content.strip()
        results[name] = ("✅", reply)
        print(f"✅ {name:15} → {reply}")
    except Exception as e:
        results[name] = ("❌", str(e))
        print(f"❌ {name:15} → ERROR: {e}")
```
