### Check if openRouter API key exists in the project

```python
import os
import sys
from dotenv import load_dotenv
from openai import OpenAI
from pathlib import Path

from src.config import LLM_MODELS

# Make sure project root is on sys.path so we can import from src/
project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")
assert api_key, "❌ OPENROUTER_API_KEY not found in .env!"
print(f"✅ API key loaded (starts with: {api_key[:10]}...)")

client = OpenAI(api_key=api_key, base_url="https://openrouter.ai/api/v1")
```

### Test if every LLM is working

```python
prompt = "Reply with exactly this and nothing else: API works!"

for name, model_id in LLM_MODELS.items():
    try:
        r = client.chat.completions.create(
            model=model_id,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=200,
        )
        reply = (r.choices[0].message.content or "").strip()
        if not reply:
            raise ValueError("empty response")
        print(f"✅ {name:18} → {reply}")
    except Exception as e:
        print(f"❌ {name:18} → ERROR: {e}")
```
